Koneksi ke MongoDB dari Python

```python
from pymongo import MongoClient

# Membuat konesi ke MongoDB yang berjalan di localhost, port default 27017
client = MongoClient("mongodb://localhost:27017/")

# Memilih (atau otomatis membuat) database
db = client["kopi_nusantara"]

# Memilih (atau otomatis membuat) collection
data_menu = db["menu"]

print("Koneksi ke MongoDB berhasil")
print("Daftar database yang ada: ", client.list_database_names())
```

A. Membangun Katalog (Create) 

```python
data_menu.delete_many({})

data_menu.insert_many([
    {"nama_menu": "Espresso", "harga": 18000, "kategori": "Kopi", "tingkat_kepahitan": 5, "stok": 30, "rekomendasi_barista":True},
    {"nama_menu": "Americano", "harga": 20000, "kategori": "Kopi", "tingkat_kepahitan": 4, "stok": 25},
    {"nama_menu": "Cappuccino", "harga": 25000, "kategori": "Kopi", "tingkat_kepahitan": 3, "stok": 20, "rekomendasi_barista":True},
    {"nama_menu": "Latte", "harga": 27000, "kategori": "Kopi", "tingkat_kepahitan": 2, "stok": 25, "rekomendasi_barista":True},
    {"nama_menu": "Kopi Susu Gula Aren", "harga": 22000, "kategori": "Kopi", "tingkat_kepahitan": 2, "stok": 40},
    {"nama_menu": "Cold Brew", "harga": 28000, "kategori": "Kopi", "tingkat_kepahitan": 4, "stok": 15, "rekomendasi_barista":True},
    {"nama_menu": "Nasi Goreng", "harga": 25000, "kategori": "Makanan", "pedas_level": 2, "stok": 20, "rekomendasi_barista":True},
    {"nama_menu": "Mie Goreng", "harga": 20000, "kategori": "Makanan", "pedas_level": 3, "stok": 25},
    {"nama_menu": "Roti Bakar", "harga": 15000, "kategori": "Makanan", "pedas_level": 0, "stok": 30},
    {"nama_menu": "Kentang Goreng", "harga": 18000, "kategori": "Makanan", "pedas_level": 0, "stok": 35},
    {"nama_menu": "Chicken Sandwich", "harga": 30000, "kategori": "Makanan", "pedas_level": 1, "stok": 15},
    {"nama_menu": "Pisang Goreng", "harga": 16000, "kategori": "Makanan", "pedas_level": 0, "stok": 25}
])

print("Jumlah menu: ", data_menu.count_documents({}))
```

B. Kueri Bertingkat (Read)

Seluruh menu kategori "Kopi" dengan tingkat_kepahitan di atas 3.

```python
for menu in data_menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}},{"_id":0}):
    print(menu)
```

Seluruh menu dengan harga di antara 15000 dan 30000.

```python
for menu in data_menu.find({"harga": {"$gte": 15000, "$lte": 30000}}, {"_id":0}):
    print(menu)
```

Menu yang tidak memiliki field pedas_level sama sekali.

```python
for menu in data_menu.find({"pedas_level": {"$exists": False}}, {"_id":0}):
    print(menu)
```

C. Pembaruan Data (Update)

```python
data_menu.update_many({},{"$set": {"status": "Tersedia"}})

data_menu.update_one(
    {"nama_menu": "Espresso"},
    {"$set": {"status": "Habis"}}
)

data_menu.update_one(
    {"nama_menu": "Nasi Goreng"},
    {"$set": {"status": "Habis"}}
)

for menu in data_menu.find({},{"_id": 0, "nama_menu": 1, "status": 1}):
    print(menu)
```

D. Integrasi ke Spark

```python
from pyspark.sql import SparkSession
import pandas as pd

spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

dokumen_menu = list(data_menu.find({}))

df_pandas = pd.DataFrame(dokumen_menu)
df_pandas["_id"] = df_pandas["_id"].astype(str)
df_pandas = df_pandas.astype(object).where(
    pd.notna(df_pandas), None
)

df_spark_menu = spark.createDataFrame(df_pandas)
df_spark_menu.printSchema()
df_spark_menu.groupBy("kategori").avg("harga").show()
```

E. Kesimpulan 

Selama proses mengubah data MongoDB menjadi Spark DataFrame, salah satu tantangan yang ditemukan adalah struktur data MongoDB 
yang tidak selalu sama pada setiap dokumen. Contohnya, menu kategori Kopi memiliki field tingkat_kepahitan, sedangkan menu 
kategori Makanan memiliki field pedas_level. Hal seperti ini membuat beberapa kolom memiliki nilai kosong atau null pada saat 
data diubah menjadi Spark DataFrame. Untuk mengatasinya, data terlebih dahulu diubah menjadi Pandas DataFrame menggunakan 
pd.DataFrame(), kemudian tipe data yang diperlukan disesuaikan sebelum dikonversi menggunakan spark.createDataFrame(). Dengan 
cara ini, data yang memiliki field berbeda tetap dapat disatukan dalam satu Spark DataFrame. Dari praktikum ini dapat 
dipahami bahwa MongoDB lebih fleksibel dalam menyimpan data, sedangkan Spark DataFrame membutuhkan struktur kolom yang lebih 
seragam agar data dapat dianalisis dengan baik.
